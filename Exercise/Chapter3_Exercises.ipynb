1. Recall that the hyperbolic sine, denoted by sinh, is $\frac{\exp(x) - \exp(-x)}{2}$. Write a function `my_sinh(x)`, where the output $y$ is the hyperbolic sine computed on $x$. Assume that $x$ is a 1 by 1 float.

**Test Cases:**
`my_sinh(0)` $\rightarrow$ `0`
`my_sinh(1)` $\rightarrow$ `1.1752`
`my_sinh(2)` $\rightarrow$ `3.6269`

```python
def my_sinh(x):
    y = (math.exp(x) - math.exp(-x)) / 2
    return round(y, 4)
my_sinh(1)
```

4. Write a function `my_split_matrix(m)`, where $m$ is an array, the output is a list `[m1, m2]` where $m1$ is the left half of $m$, and $m2$ is the right half of $m$. In the case where there is an odd number of columns, the middle column should go to $m1$. Assume that $m$ has at least two columns.

```python
def my_split_matrix(m):
    r, c = m.shape

    m1 = m[:, 0:(c + 1) // 2]
    m2 = m[:, (c + 1) // 2:]

    return [m1, m2]

m = np.array([[1, 2,3], 
              [4, 5, 6]])
print(my_split_matrix(m))
```

5. Write a function `my_cylinder(r, h)`, where $r$ and $h$ are the radius and height of a cylinder, respectively, and the output is a list `[s, v]` where $s$ and $v$ are the surface area and volume of the same cylinder, respectively. Recall that the surface area of a cylinder is $2\pi r^2 + 2\pi rh$, and the volume is $\pi r^2 h$. Assume that $r$ and $h$ are 1 by 1 float.

```python
def my_cylinder(r, h):
    s = 2 * math.pi * pow(r, 2) + 2 * math.pi * r * h
    v = math.pi * pow(r, 2) * h
    return [s, v]
```

6. Write a function `my_n_odds(a)`, where $a$ is a one-dimensional array of floats and the output is the number of odd numbers in $a$.

```python
def my_n_odds(a):
    a = a[a % 2 != 0]

    return a.size

my_n_odds(np.array([1, 2, 3, 4, 5]))
```

12. Let $r1$ and $r2$ be the radius of circles with the same center and let $r2>r1$. Write a function `my_donut_area(r1, r2)`, where the output is the area outside of the circle with radius $r1$ and inside the circle with radius $r2$. Make sure that the function is vectorized. Assume that $r1$ and $r2$ are one-dimensional array of the same size.

```python
def my_donut_area(r1, r2):
    return math.pi * (r2 ** 2 - r1 ** 2)
```

13. Write a function `my_within_tolerance(A, a, tol)`, where the output is an array or list of the indices in $A$ such that $\vert{}A - a\vert{} < tol$. Assume that $A$ is a one-dimensional float list or array and that $a$ and $tol$ are 1 by 1 floats.

```python
def my_within_tolerance(A, a, tol):
    indices = []
    for i, val in enumerate(A):
        if abs(val - a) < tol:
            indices.append(A[i])
    return indices
my_within_tolerance([0, 1, 2, 3], 1.5, 0.75)
```

14. Write a function `bounding_array(A, top, bottom)`, where the output is equal to the array $A$ wherever $bottom < A < top$, the output is equal to $bottom$ wherever $A <= bottom$, and the output is equal to $top$ wherever $A >= top$. Assume that A is one-dimensional float array and that $top$ and $bottom$ are 1 by 1 floats.

```python
def bounding_array(A, top, bottom):
    output = []
    for val in A:
        if val <= bottom:
            output.append(bottom)
        elif val >= top:
            output.append(top)
        else:
            output.append(int(val))
    return output

bounding_array(np.arange(-5, 6, 1), 3, -3)
```
