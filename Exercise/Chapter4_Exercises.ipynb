1. Write a function `my_tip_calc(bill, party)` where `bill` is the total cost of a meal and `party` is the number of people in the group. The tip should be calculated as 15% for a party strictly less than six people, 18% for a party strictly less than eight, 20% for a party less than 11, and 25% for a party 11 or more.<br><br>**Test Cases:**<br>`my_tip_calc(109.29, 3)` $\rightarrow$ `16.3935`<br>`my_tip_calc(109.29, 7)` $\rightarrow$ `19.6722`<br>`my_tip_calc(109.29, 9)` $\rightarrow$ `21.8580`<br>`my_tip_calc(109.29, 12)` $\rightarrow$ `27.3225`

```python
def my_tip_calc(bill, party):
    if party < 6:
        rate = 0.15
    elif party < 8:
        rate = 0.18
    elif party < 11:
        rate = 0.2
    else:
        rate = 0.25

    return round(bill * rate, 4)
my_tip_calc(109.29, 9)
```

2. Write a function `my_mult_operation(a,b,operation)`. The input argument, `operation`, is a string that is either \"plus\", \"minus\", \"mult\", \"div\", or \"pow\", and the function should compute: $a+b$, $a-b$, $a \times b$, $a/b$, and $a^b$ for the respective values for operation.<br><br>**Test Cases:**<br>`x = np.array([1,2,3,4])`<br>`y = np.array([2,3,4,5])`<br>`my_mult_operation(x, y, 'plus')` $\rightarrow$ `[3, 5, 7, 9]`<br>`my_mult_operation(x, y, 'minus')` $\rightarrow$ `[-1, -1, -1, -1]`<br>`my_mult_operation(x, y, 'mult')` $\rightarrow$ `[2, 6, 12, 20]`<br>`my_mult_operation(x, y, 'div')` $\rightarrow$ `[0.5, 0.66666667, 0.75, 0.8]`<br>`my_mult_operation(x, y, 'pow')` $\rightarrow$ `[1, 8, 81, 1024]`

```python
def my_mult_operation(a, b, operation):
    if operation == 'plus':
        return a + b
    elif operation == 'minus':
        return a - b
    elif operation == 'mult':
        return a * b
    elif operation == 'div':
        return a / b
    else:
        return a ** b
x = np.array([1, 2, 3, 4])
y = np.array([2, 3, 4, 5])
my_mult_operation(x, y, 'plus')
```

3. Consider a triangle with vertices at $(0,0)$, $(1,0)$, and $(0, 1)$. Write a function with the name `my_inside_triangle(x,y)` where the output is the string 'outside' if the point $(x, y)$ is outside of the triangle, 'border' if the point is exactly on the border of the triangle, and 'inside' if the point is on the inside of the triangle.<br><br>**Test Cases:**<br>`my_inside_triangle(.5, .5)` $\rightarrow$ `'border'`<br>`my_inside_triangle(.25, .25)` $\rightarrow$ `'inside'`<br>`my_inside_triangle(5, 5)` $\rightarrow$ `'outside'`

```python
def my_inside_triangle(x, y):
    if x < 0 or y < 0 or (x + y) > 1:
        return 'outside'
    elif x == 0 or y == 0 or (x + y) == 1:
        return 'border'
    else:
        return 'inside'
my_inside_triangle(0.5, 0.5)
```
